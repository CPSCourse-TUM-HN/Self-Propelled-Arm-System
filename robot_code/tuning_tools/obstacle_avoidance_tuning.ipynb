# Simplified TangentBug Obstacle Tuning

This standalone tool detects center-path depth obstacles without requiring a specific object shape, draws bbox and tangent guides, performs a symmetric bypass, and approximately restores the original route and heading. `perception.obstacle_top_ignore_ratio` and `obstacle_bottom_ignore_ratio` affect obstacle bboxes only; camera frames, DepthNet input, depth baselines, HUD, and recordings remain full-size. Geometry HUD and Text HUD can be toggled independently; detailed reasons and motion estimates remain in logs. The same center-corridor bbox must persist for `control.obstacle_confirmation_frames` stationary observations before it is locked. `control.linear_obstacle_confirmation_frames` applies stop-and-confirm to new center-path bboxes seen during OFFSET_OUT, PASSING, and OFFSET_RETURN; confirmed bboxes start a nested bypass, and interrupted movement is retained for route restoration. During TURN_TO_TANGENT, a frame is safe when the locked bbox is outside the collision corridor or absent; `control.tangent_safe_frames` consecutive safe frames complete the turn. Frames classified as IGNORED_WALL are clear-path samples and increment this streak, while absolute-depth protection remains active. `motion.depth_to_segment_time_factor` controls the first lateral segment and symmetric return; `motion.pass_depth_to_time_factor` independently controls the deterministic second straight segment. Parameters are read only from `obstacle_avoidance_parameters.json`. Start with `base_real` disabled and use **Analyze Frame** before real motion.
