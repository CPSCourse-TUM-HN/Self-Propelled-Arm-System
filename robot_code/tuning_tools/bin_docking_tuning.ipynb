```python
from pathlib import Path
import sys

def find_project_root():
    for candidate in (Path.cwd(), *Path.cwd().parents):
        if (candidate / 'config.json').is_file():
            return candidate
    raise RuntimeError('project root containing config.json was not found')

PROJECT_ROOT = find_project_root()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from IPython.display import display
from tuning_tools.bin_docking_tuning import BinDockingTuningPanel

panel = BinDockingTuningPanel()
display(panel.widget())
print('[bin-docking] ready; use STOP ALL before changing hardware placement')
```
